# ATML PA4 public data prep

CPU-only. Clones the pinned course data repo and runs `setup_data.py --public`.
Does not download model weights and does not build prefix banks.

```python
import shutil
import subprocess
import sys
from pathlib import Path

COMMIT = "6a0a9ec9da7ef3d313ee3036cfec493886723644"
root = Path("/kaggle/working/atml-pa4-datasets")
if root.exists():
    shutil.rmtree(root)
subprocess.check_call(["git", "clone", "https://github.com/AbDu11aHHH/atml-pa4-datasets.git", str(root)])
subprocess.check_call(["git", "-C", str(root), "checkout", "--detach", COMMIT])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "-r", str(root / "requirements-public.txt")])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "transformers==4.57.1", "huggingface_hub==0.36.2"])
subprocess.check_call([sys.executable, str(root / "scripts" / "setup_data.py"), "--public"], cwd=root)
subprocess.check_call([sys.executable, str(root / "scripts" / "check_setup.py")], cwd=root)
bundle = Path("/kaggle/working/pa4_public_locked")
if bundle.exists():
    shutil.rmtree(bundle)
shutil.copytree(root / "data" / "public_locked", bundle / "public_locked")
shutil.copy(root / "data" / "local_source_revisions.json", bundle / "local_source_revisions.json")
print("bundle ready", bundle)
for path in sorted(bundle.rglob("*")):
    if path.is_file():
        print(f"{path.relative_to(bundle)} {path.stat().st_size:,}")
```
