# ATML PA4 Task 2 MQAR training

All 42 training jobs (14 architectures x 3 seeds), then analyze.

```python
import gzip, os, shutil, subprocess, sys
from pathlib import Path

root = Path('/kaggle/input')
print('inputs', [p.name for p in root.iterdir()], flush=True)
locked = next(root.rglob('c4_calibration_locked.jsonl*'))
source = locked.parents[2]
data = Path('/kaggle/working/pa4-data')
if data.exists():
    shutil.rmtree(data)
gzip_dirs = {'public_locked', 'kv_retrieval', 'mqar'}
for path in source.rglob('*'):
    if not path.is_file():
        continue
    rel = path.relative_to(source)
    if path.suffix == '.jsonl' and path.parent.name in gzip_dirs:
        target = data / rel.with_suffix('.jsonl.gz')
        target.parent.mkdir(parents=True, exist_ok=True)
        with path.open('rb') as raw, gzip.open(target, 'wb') as packed:
            shutil.copyfileobj(raw, packed)
    else:
        target = data / rel
        target.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy2(path, target)
print('data', data, flush=True)

repo = Path('/kaggle/working/ATML-PA4')
if repo.exists():
    shutil.rmtree(repo)
subprocess.check_call(['git', 'clone', 'https://github.com/ttqureshi/ATML-PA4.git', str(repo)])
subprocess.check_call(['git', '-C', str(repo), 'checkout', '--detach', 'origin/task2'])
subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'transformers==4.57.1', 'huggingface_hub==0.36.2', 'accelerate', 'matplotlib'])

env = os.environ.copy()
env['PA4_DATA_REPO'] = str(data)
env['PYTHONUNBUFFERED'] = '1'
env['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'

for stage_args in [
    ['task2/run.py', 'train'],
    ['task2/run.py', 'analyze'],
]:
    print('\n====', ' '.join(stage_args), '====', flush=True)
    rc = subprocess.run([sys.executable, *stage_args], cwd=repo, env=env).returncode
    print('exit', rc, flush=True)
    if rc != 0:
        raise SystemExit(rc)

bundle = Path('/kaggle/working/task2_results')
if bundle.exists():
    shutil.rmtree(bundle)
src = repo / 'results' / 'task2'
if src.exists():
    shutil.copytree(src, bundle)
ckpt = repo / 'checkpoints'
if ckpt.exists():
    shutil.copytree(ckpt, Path('/kaggle/working/task2_checkpoints'))
for bulky in (repo, data, Path('/kaggle/working/.cache'), Path('/root/.cache')):
    if bulky.exists():
        shutil.rmtree(bulky, ignore_errors=True)
print('TASK2 TRAIN COMPLETE', flush=True)
```
